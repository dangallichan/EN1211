## **Task 4** - Introducing 'functions' to simplify the readability of code

You are now given starting code below which defines a function called 'drawT()' to draw the letter 'T'.
Run the cell below, and see that **it doesn't actually appear to do anything at the moment!**

```python

# Define a function called 'drawT()'
def drawT(t,letterSize):
    # Store the current heading and position of the turtle
    startHead = t.heading
    startX = t.x
    startY = t.y

    # Moves the turtle around to draw the letter T
    t.forward(10*letterSize)
    t.left(90)
    t.forward(10*letterSize/2)
    t.right(180)
    t.forward(10*letterSize)

    # Jump the turtle back to the original position and heading
    t.heading = startHead
    t.jump_to(startX,startY)



```

The code above defines the function, but if we want to use that code we need to 'call' our function to actually run it:

```python
# Run this cell to draw the T again
t = Turtle()
t.pen_color = 'orchid'
letterSize = 10
drawT(t,letterSize)
```

## **Task 6** - From Turtles to Car Wheels

<img src="data:image/svg+xml;utf8,<svg xmlns='http://www.w3.org/2000/svg' width='800' height='10'><rect width='100%' height='100%' fill='%23f8f9fa' rx='6'/></svg>" width="100%" />


There are lots of different designs for car wheels (which seem to often be about aesthetics rather than strength!). Here are some examples:


carWheels.jpg


Start by running the cell below, and hopefully you should see that the turtle draws a basic wheel design.

You can verify that it is really the turtle doing the work by increasing the `t.delay` parameter (to something like `0.2`). You can always put this back to zero to speed it up again if you want.


```python
# Make a wheel design with our turtle!

import math    # We will need some trigonometry operations for this!

t = Turtle() # Create a new turtle

# Set some basic parameters for the design of our wheel
wheelRadius = 160
tyreThickness = 30
hubRadius = 30

# Draw the tyre and hub
t.addWheel(wheelRadius+tyreThickness/2,hubRadius,tyreThickness)

t.pen_width = 4
t.pen_color = "rgb(128,128,128)"    # Here specifying colour using R,G,B components between 0 and 255 - see https://www.w3.org/TR/2003/REC-SVG11-20030114/types.html#ColorKeywords
t.delay = 0           # Go as fast as possible
t.visible = False     # We don't actually need to see our turtle


# Set some parameters for the wheel spokes
midRadius = 80
Angle1 = 30
Angle2 = 40
nSpokes = 8

# Now draw the spokes
for iSpoke in range(nSpokes):

    theta = iSpoke * 2.0 * math.pi / nSpokes   # find the starting angle for this spoke in radians

    # Move to starting position for this spoke (without drawing)
    t.jump_to(hubRadius*math.sin(theta),hubRadius*math.cos(theta))
    t.heading = (-theta * 180.0 / math.pi) + 90 # also need to convert to degrees here!

    # draw the first side of the current spoke
    t.left(Angle1)
    t.forwardUntilRadius(midRadius) # Keeps going in current direction until this radius is reached
    t.right(Angle2)
    t.forwardUntilRadius(wheelRadius)

    # move back to starting position for this spoke (without drawing)
    t.jump_to(hubRadius*math.sin(theta),hubRadius*math.cos(theta))
    t.heading =(-theta * 180.0 / math.pi) + 90 # also need to convert to degrees here!

    # draw the second side
    t.right(Angle1)
    t.forwardUntilRadius(midRadius) # Keeps going in current direction until this radius is reached
    t.left(Angle2)
    t.forwardUntilRadius(wheelRadius)
```

In this code, the wheel has been parameterised according to the following diagram:


wheelParameterisation.jpg


The 3 parameters in the diagram correspond to `Angle1`, `Angle2` and `midRadius` in the code. There is also a 4th parameter, `nSpokes`, which sets the number of spokes that are drawn.
